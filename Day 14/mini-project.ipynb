```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, RidgeCV, LassoCV
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import root_mean_squared_error, r2_score
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from collections import Counter
```

### Loading and EDA on the dataset, finding out it's charateristics. 

```python
df = pd.read_csv('CleanAmes.csv')
```

```python
# Tests
pd.set_option('display.max_columns', 100)
df.head(2)
```

```python
df['age'] = 2006 - df['Year Built']
df['from_remod'] = 2006 - df['Year Remod/Add']
df.drop(columns=['Year Built', 'Year Remod/Add'], inplace=True)
df = df.dropna()
X = df.drop(['SalePrice', 'Sale Type'], axis=1)
y = np.log1p(df['SalePrice'])
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
```

```python
enc_cols = ['Neighborhood', 'Exterior 1st', 'Exterior 2nd', 'MS SubClass']
scl_cols = ['Lot Frontage', 'Lot Area', 'Mas Vnr Area', 'BsmtFin SF 1', 'BsmtFin SF 2', 'Bsmt Unf SF', '1st Flr SF', '2nd Flr SF', 'Low Qual Fin SF', 'Gr Liv Area', 'Garage Yr Blt', 'Garage Area', 'Wood Deck SF', 'Open Porch SF']

encode = Pipeline([
    ('encode', OneHotEncoder(sparse_output=False, handle_unknown='ignore'))
])

scale = Pipeline([
    ('ss', StandardScaler())
])

prep = ColumnTransformer([
    ('enc', encode, enc_cols),
    ('scl', scale, scl_cols)
],
    remainder='passthrough')

pipeline = Pipeline([
    ('prep', prep), 
    ('lr', RidgeCV(alphas=np.array(range(5)), cv=5))
])
```

```python

pipeline.fit(X_train, y_train)
pred = pipeline.predict(X_test)
r2 = r2_score(y_test, pred)
scores = cross_val_score(pipeline, X, y, cv=5)
```
