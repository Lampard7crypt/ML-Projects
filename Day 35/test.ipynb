# This file Contains the Ames Housing joblib importation to see if everything works alright. 

```python
import joblib
import pandas as pd
import numpy as np

# Run ames_project.ipynb first to get the .joblib file. 
loaded_model = joblib.load('ames_housing_model.joblib') 
path = "../Data/AmesHousing.csv"
df = pd.read_csv(path)

# Must be same schema as the data fitted, in this case the only time the df was affected was during feature engineering. 
df['Was_remodeled'] = (df['Year Remod/Add'] != df['Year Built']).astype(int)
df['age'] = 2026 - df['Year Built']
df = df.drop(columns=['Year Built', 'Year Remod/Add']) # two columns added and 2 dropped. 
# ------------------------------------------------------------------
X = df.drop('SalePrice', axis=1)
y = np.log1p(df['SalePrice'])
loaded_model.score(X, y)
```
